```python
# import necessary packages
import os
import pandas as pd
import pym
import numpy as np
import openpyxl
```

```python
fp_temp_change = r'GWL\Temp_Est.out'

# load mym output file as array of tuples
def load_files(fp):
    file = pym.read_mym(fp)
    return file

temp_change_tuple = load_files(fp_temp_change)
```

```python
def reshape_data(data_tuple):
    
    # Unpack the tuple
    data, years = data_tuple

    # Flatten the data array if it is not 1D
    data = np.ravel(data)

    # Create a DataFrame
    df = pd.DataFrame({
        'year': years,
        'gwl': data
    })
    
    return df

temp_change = reshape_data(temp_change_tuple)
```

```python
def add_variables(df):
    
    df.insert(0, "Unit", "Degrees Celcius", allow_duplicates = False)
    df.insert(0, "Variable", "Temperature Change", allow_duplicates = False)
    df.insert(0, "Region", "World", allow_duplicates = False)
    df.insert(0, "Scenario", "SSP2", allow_duplicates = False)
    df.insert(0, "Model", "IMAGE 3.4", allow_duplicates = False)

    return df

temp_change = add_variables(temp_change)

# repeat row as having only on scenario leads to error in rime 
temp_change_new = pd.DataFrame(np.repeat(temp_change.values, 2, axis = 0))
temp_change_new.columns = temp_change.columns
```
